##### Identify Negative sales amounts 

```python
%sql
select * from transactions
where Total_Amount < 0
```

##### Identify Zero value transactions 

```python
%sql
select * from transactions where Total_Amount = 0
```

##### 	Duplicate transactions 

```python
%sql
select Transaction_ID, count(1)
from transactions
group by Transaction_ID
having count(Transaction_ID) > 1
```

##### Unusual spikes/drops 

```python
%sql
with stats as 
(
    select avg(total_amount) as avg_val, 
    stddev(total_amount) as std_dev
    from transactions
)
select 
    transaction_id, 
    total_amount, 
    case
        when ABS((total_amount-avg_val)/std_dev) > 3 then 'Anamoly'
        else 'Normal'
    end as anamoly_flag
from stats cross join transactions
```

##### Unrealistic quantities 

```python
%sql
with stats as 
(
    select avg(Quantity) as avg_qty,
    stddev(Quantity) as std_qty
    from transactions
),
transactions_details as
(select 
    transaction_id, 
    Quantity, 
    case
        when ABS((Quantity-avg_qty)/std_qty) > 3 then 'Anamoly'
        else 'Normal'
    end as anamoly_flag
from stats cross join transactions)

select * from transactions_details
where anamoly_flag = 'Anamoly'
```

##### Invalid loyalty redemptions → points redeemed without a valid account.

```python
%sql
select * from transactions
where Loyalty_Redeemed > 0 and Customer_ID is null
```

##### •	Transactions outside store hours → sales logged when store is closed. 10PM is closing time

```python
%sql
SELECT *
FROM transactions
WHERE CAST(Time AS TIME) > CAST('22:00:00' AS TIME);
```

```python
%sql
-- Reference table of valid SKU mappings
WITH sku_master AS (
    SELECT 'SKU1001' AS SKU, 'Soda 500ml' AS Product_Name UNION ALL
    SELECT 'SKU2002', 'Chips 100g' UNION ALL
    SELECT 'SKU3003', 'Chocolate Bar' UNION ALL
    SELECT 'SKU4004', 'Milk 1L' UNION ALL
    SELECT 'SKU5005', 'Bread Loaf'
)
SELECT 
    t.Transaction_ID,
    t.SKU,
    t.Product_Name,
    CASE 
        WHEN m.Product_Name IS NULL THEN 'Unknown SKU'
        WHEN t.Product_Name <> m.Product_Name THEN 'Mismatch'
        ELSE 'Valid'
    END AS anomaly_flag
FROM transactions t
LEFT JOIN sku_master m
    ON t.SKU = m.SKU;
```
